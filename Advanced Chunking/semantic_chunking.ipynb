```python
from sentence_transformers import SentenceTransformer
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

## Sample text
text = """
Langchain is a framework for developing applications powered by language models. It can be used for chatbots, Generative Question-Answering (GQA), summarization, and much more.
It also provides integrations with other tools to help you build your application. 
You can create chains, agents, memory and retrievers using Langchain. Chains are a way to combine multiple LLM calls into a single call. Agents are a way to use LLMs to decide what to do next. Memory is a way to store information between calls. Retrievers are a way to get information from external sources.
Eiffel Tower is a wrought-iron lattice tower on the Champ de Mars in Paris, France. 
Paris is the capital and most populous city of France.
"""
```

```python
sentences = [s.strip() for s in text.split('\n') if s.strip()]
#step 2: Embed each sentence using the model
embeddings = model.encode(sentences)
# Step 3: Initialize parameters for chunking
threshold = 0.5
chunks = []
current_chunk = [sentences[0]]
# Step 4: Semantic grouping of sentences based on cosine similarity and threshold

for i in range(1, len(sentences)):
    # Calculate cosine similarity between the current sentence and the last sentence in the current chunk
    similarity = cosine_similarity([embeddings[i-1]], [embeddings[i]])[0][0]
    
    if similarity >= threshold:
        current_chunk.append(sentences[i])
    else:
        chunks.append(" ".join(current_chunk))
        current_chunk = [sentences[i]]

# Append the last chunk
chunks.append(" ".join(current_chunk))
```

```python
# output the chunks
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {chunk}\n")
```
