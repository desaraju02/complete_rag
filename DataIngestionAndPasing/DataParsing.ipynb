```python
# Create a simple text file
import os
os.makedirs("data/text_files", exist_ok=True)
```

```python
sample_texts = {
    "data/text_files/rag_intro.txt": """
🔍 What Is RAG?
Retrieval‑Augmented Generation (RAG) is an AI architecture that enhances large language models by allowing them to fetch relevant external information before generating a response. Instead of relying solely on what the model learned during training, RAG supplements the prompt with real, retrieved knowledge, making answers more accurate and grounded.
🧠 How RAG Works
A RAG system typically combines two components: a retriever and a generator. The retriever searches a knowledge base—often a vector database—to find the most relevant documents or text chunks. These retrieved pieces are then passed to the generator (usually an LLM), which uses them to craft a coherent, context‑aware response. This pipeline reduces hallucinations and improves factual reliability.
🚀 Why RAG Matters
RAG is especially powerful for domains where information changes frequently or requires deep specialization, such as finance, healthcare, legal research, or enterprise knowledge systems. It allows organizations to keep AI responses up‑to‑date without retraining the entire model, making it cost‑efficient and scalable.
📚 Real‑World Impact
From customer support bots that reference internal manuals to research assistants that pull scientific papers on demand, RAG is reshaping how AI interacts with knowledge. By blending retrieval with generation, it creates systems that are not only smarter but also more trustworthy and adaptable.

    """,
}
```

```python
for file_path, content in sample_texts.items():
    with open(file_path, "w") as f:
        f.write(content.strip())
```

```python
# Text Loader - Single File
from langchain_community.document_loaders import TextLoader

loader = TextLoader("data/text_files/rag_intro.txt")
documents = loader.load()
print(f"Loaded {len(documents)} document(s) from the text file.")
```

```python
doc = documents[0]
print(doc.page_content)
print(doc.metadata)
```

```python
print(f"Content Preview: {doc.page_content[:100]}...")
```

```python
## Read all text files in a directory
from langchain_community.document_loaders import DirectoryLoader
directory_loader = DirectoryLoader("data/text_files", glob="**/*.txt", loader_cls=TextLoader, show_progress=True)
all_documents = directory_loader.load()
print(f"Loaded {len(all_documents)} document(s) from the directory.")
```

```python
print(f"Loaded {len(all_documents)} document(s) from the directory.")
for i, doc in enumerate(all_documents):
    print(f"\nDocument {i+1}:")
    print(f"Metadata: {doc.metadata}")
    print(f"Content Preview: {doc.page_content[:100]}...")
    print(f"Length of content: {len(doc.page_content)} characters")
```

### Text Splitting Strategies


```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, CharacterTextSplitter, TokenTextSplitter

```

```python
# Method 1 - CharacterTextSplitter
char_splitter = CharacterTextSplitter(separator="\n", chunk_size=100, chunk_overlap=20, length_function=len)
char_chunks = char_splitter.split_text(documents[0].page_content)
print(f"Character Splitter produced {len(char_chunks)} chunks:")
for i, chunk in enumerate(char_chunks):
    print(f"\nChunk {i+1}:")
    print(f"Content: {chunk}")
    print(f"Length: {len(chunk)} characters")
```

### Recursive Character text splitter

```python
print(" This is recursive character text splitter")
recursive_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", " ", ""],
    chunk_size=100,
    chunk_overlap=20,
    length_function=len
)
recursive_chunks = recursive_splitter.split_text(documents[0].page_content)
print(f"Recursive Character Splitter produced {len(recursive_chunks)} chunks:")
for i, chunk in enumerate(recursive_chunks):
    print(f"\nChunk {i+1}:")
    print(f"Content: {chunk}")
    print(f"Length: {len(chunk)} characters")
```
